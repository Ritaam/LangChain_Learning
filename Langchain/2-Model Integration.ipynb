##### Streaming  and Batch

Streaming 

Most Model can stream their output  content while it is being generated. By displaying output progressively , streaming user exprience, particularly  for longer responses. calling stream() returns  an iterator that  yeilds output chunks as they are produced. we can use a loop to process each chunk in real time.

```python
model.invoke("write me a 200 words paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Write me a 200 words paragraph in deep learning"):
    print(chunk.text)
```

### Batch

Batcging a collection of independent request to a model can significantly improve performance and reduce cost , as the processing can be done in parallel

```python
responses = model.batch(["what is Machine Learning?" , 
   "how do airplane fly?", 
   "What is quantam computing ?",
   "what is Cloud Computing ?"
])
   
for response in responses:
    print(response)
```
